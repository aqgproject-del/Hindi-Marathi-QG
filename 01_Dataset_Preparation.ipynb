```python
import os
import re
import unicodedata
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
input_path = "/content/drive/MyDrive/AQG/extracted_text.txt"

with open(input_path, "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

```python
import re

text = re.sub(r'\s+', ' ', text)
text = text.strip()

print(text)
```

```python
import unicodedata

text = unicodedata.normalize("NFC", text)

print(text)
```

```python
output_path = "/content/drive/MyDrive/AQG/clean_text.txt"

with open(output_path, "w", encoding="utf-8") as file:
    file.write(text)

print("unicodenormalization done")
```
