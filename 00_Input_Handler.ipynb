```python
import os
import fitz
from docx import Document

from google.colab import drive
```

```python
drive.mount('/content/drive')
input_file = "/content/drive/MyDrive/chapter.docx"
```

```python
file_extension = os.path.splitext(input_file)[1].lower()

print("Detected File Type :", file_extension)
```

```python
def extract_txt(path):

    with open(path, "r", encoding="utf-8") as file:
        return file.read()
```

```python
def extract_pdf(path):

    document = fitz.open(path)

    text = ""

    for page in document:
        text += page.get_text()

    return text
```

```python
def extract_docx(path):

    document = Document(path)

    text = ""

    for para in document.paragraphs:
        text += para.text + "\n"

    return text
```

```python
if file_extension == ".txt":

    text = extract_txt(input_file)

elif file_extension == ".pdf":

    text = extract_pdf(input_file)

elif file_extension == ".docx":

    text = extract_docx(input_file)

elif file_extension == ".doc":

    raise ValueError("Please convert .doc to .docx before processing.")

else:

    raise ValueError("Unsupported File Type")
```

```python
output_path = "/content/drive/MyDrive/AQG/extracted_text.txt"

with open(output_path, "w", encoding="utf-8") as file:
    file.write(text)

print("✅ extracted_text.txt created successfully!")
```
